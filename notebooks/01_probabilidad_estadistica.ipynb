```python
# Actividad 2 - Código

import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/annual-population-growth/annual-population-growth.csv")

print("-- Primeras filas:\n", df.head())
print("\n-- Dimensiones (filas, columnas):", df.shape)
print("\n-- Nombres de columnas:", df.columns.tolist())
print("\n-- Tipos de datos:\n", df.dtypes)
print("\n-- Descripciones:", df.describe())
print("\n-- Valores faltantes por columna:", df.isnull().sum())
print("\n-- Cantidad de entidades distintas:", df["Entity"].nunique())

# Descripción de lo realizado: Importar csv y usar funciones como head, shape, tolist, etc. para ver datos generales sobre el acomodo, tipos y cantidad de los datos del csv
```

```python
# Actividad 3 - Código
anio = 2015
df_anio = df[df["Year"] == anio] #filtrar por año
cambio_poblacion = df_anio["Annual change in population"] #asignar los datos de la columna a una variable

print("-- Media:", cambio_poblacion.mean())
print("-- Mediana:", cambio_poblacion.median())
print("-- Varianza:", cambio_poblacion.var())
print("-- Desviación estándar:", cambio_poblacion.std())
print("-- Mínimo:", cambio_poblacion.min())
print("-- Máximo:", cambio_poblacion.max())
print("\n-- Entidad con el valor mínimo:\n", df_anio[cambio_poblacion == cambio_poblacion.min()])
print("\n-- Entidad con el valor máximo:\n", df_anio[cambio_poblacion == cambio_poblacion.max()])

# Descripción de lo realizado: usar funciones como mean, median, etc. para calcular datos en base a una una columna obtenida de un año específico
```
